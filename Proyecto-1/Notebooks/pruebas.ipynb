```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://www.mercadona.es"
```

```python
request = requests.get(url)
```

```python
print(request.text)
```

```python
soup = BeautifulSoup(request.text, 'html.parser')
```

```python
box_codigo_postal = soup.find('input', {'aria-label': 'Código postal', 'maxlength': '5'})
print(box_codigo_postal)
```
