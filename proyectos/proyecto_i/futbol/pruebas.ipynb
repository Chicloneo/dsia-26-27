```python
from pathlib import Path
import pandas as pd
from pandas import DataFrame
```

```python
ruta = "datos/datos_transfermarkt/players.csv"
df = pd.read_csv(ruta)
df.head()
```

```python
def eliminar_columnas(df: DataFrame, lista_columnas: list[str]) -> DataFrame:
    return df.drop(lista_columnas, axis=1)

df = eliminar_columnas(df, ["image_url", "url", "first_name", "last_name", "name"])
df.head()
```
